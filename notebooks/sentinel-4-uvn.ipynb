# Sentinel-4 / UVN (Ultraviolet Visible Near-infrared spectrometer)

Example notebook from the [Satellite Data Catalog](https://alessandrosebastianelli.github.io/satcat/satellite.html?folder=sentinel/sentinel-4-uvn). Each section below runs the Python example of one processing level.

Run the cells top to bottom. Some sources need a free account; the cell that needs credentials asks for them when it runs.

## L1B

Level 1B: calibrated, geolocated radiance spectra and solar irradiance (hourly, east-west scans).

Looks for Sentinel-4 collections in the EUMETSAT Data Store and downloads the first product of the window. Requires a free Data Store account; collections appear once the data are released.

```python
%pip install -q eumdac satpy netCDF4 h5netcdf matplotlib
```

```python
import datetime as dt
import getpass
import shutil
import zipfile
from pathlib import Path

import eumdac

# Free consumer key / secret: https://api.eumetsat.int/api-key (EUMETSAT Data Store account)
token = eumdac.AccessToken((input("EUMETSAT consumer key: "), getpass.getpass("EUMETSAT consumer secret: ")))
datastore = eumdac.DataStore(token)
collections = [c for c in datastore.collections if "Sentinel-4" in str(c.title)]
print([(c.collection_id, str(c.title)) for c in collections])
if not collections:
    raise SystemExit("Collection not found in the Data Store yet - check data.eumetsat.int")
collection = collections[0]

products = list(collection.search(dtstart=dt.datetime(2026, 6, 1, 10, 0), dtend=dt.datetime(2026, 6, 1, 12, 0)))
print(len(products), "Sentinel-4 products found")
if not products:
    raise SystemExit("Nothing in this time window - try another one")
product = products[0]
print(product)

out = Path("data")
out.mkdir(exist_ok=True)
with product.open() as src, open(out / src.name, "wb") as dst:
    shutil.copyfileobj(src, dst)
for z in out.glob("*.zip"):                     # some collections arrive zipped
    with zipfile.ZipFile(z) as zf:
        zf.extractall(out / z.stem)
print(list(out.rglob('*.nc'))[:5])
```
