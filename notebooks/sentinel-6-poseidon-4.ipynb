# Sentinel-6 / Poseidon-4 (radar altimeter)

Example notebook from the [Satellite Data Catalog](https://alessandrosebastianelli.github.io/satcat/satellite.html?folder=sentinel/sentinel-6-poseidon-4). Each section below runs the Python example of one processing level.

Run the cells top to bottom. Some sources need a free account; the cell that needs credentials asks for them when it runs.

## L1

Level 1: calibrated altimeter waveforms (L1A and L1B, high-resolution and low-resolution modes).

Lists the Sentinel-6 collections of the EUMETSAT Data Store and downloads the first product of the window. Requires a free Data Store account; check the printed list and pick the collection (L1 or L2) you need.

```python
%pip install -q eumdac satpy netCDF4 h5netcdf matplotlib
```

```python
import datetime as dt
import getpass
import shutil
import zipfile
from pathlib import Path

import eumdac

# Free consumer key / secret: https://api.eumetsat.int/api-key (EUMETSAT Data Store account)
token = eumdac.AccessToken((input("EUMETSAT consumer key: "), getpass.getpass("EUMETSAT consumer secret: ")))
datastore = eumdac.DataStore(token)
collections = [c for c in datastore.collections if "Sentinel-6" in str(c.title)]
print([(c.collection_id, str(c.title)) for c in collections])
if not collections:
    raise SystemExit("Collection not found in the Data Store yet - check data.eumetsat.int")
collection = collections[0]

products = list(collection.search(dtstart=dt.datetime(2024, 6, 1, 0, 0), dtend=dt.datetime(2024, 6, 2, 0, 0)))
print(len(products), "Sentinel-6 products found")
if not products:
    raise SystemExit("Nothing in this time window - try another one")
product = products[0]
print(product)

out = Path("data")
out.mkdir(exist_ok=True)
with product.open() as src, open(out / src.name, "wb") as dst:
    shutil.copyfileobj(src, dst)
for z in out.glob("*.zip"):                     # some collections arrive zipped
    with zipfile.ZipFile(z) as zf:
        zf.extractall(out / z.stem)
print(list(out.rglob('*.nc'))[:5])
```
